```python
import numpy as np
import random 
```

```python
l = [1, 2, 3, 4, 5, 6, 7]
array = np.array(l)
```

```python
print(array)
```

```python
l = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
array = np.array(l)
print(array)
```

```python
array = np.zeros((4, 2), dtype = int)
print(array)
```

```python
array = np.ones((2, 2), dtype = int)
print(array)
```

```python
array = np.full((3, 3), 4, dtype = int)
print(array)
```

```python
array = np.identity(5, dtype = int)
print(array)
```

```python
array = np.linspace(1, 10, 3, dtype=int)
print(array)
```

```python
array = np.random.randint(1, 100, (4, 4))
print(array)
```
