# **Indexing and Slicing**

```python
import numpy as np
```

```python
array = np.arange(1, 10).reshape((3, 3))
print(array)
```

```python
print(array[0 ,0])
```

```python
print(array[-1, -1])
```

```python
print(array[0, 2])
```

```python
array = np.arange(1, 12)
print(array)
```

```python
print(array[: : -1])
```

```python
print(array[0:5])
```

```python
print(array[: -6: -1])
```

```python
print(array[: : 2])
```

```python
array = np.arange(1, 10).reshape((3, 3))
print(array)
```

```python
array[0, :]
```

```python
array[:, 0]
```

```python
array[0:2, ]
```

```python
array[:, 0:2]
```

```python
array[0:2, 0:2]
```

```python
array[2, 2] = 10
```

```python
print(array)
```

```python
array[0, :] = 0
```

```python
print(array)
```

# **Basic Operations**

```python
num1 = np.arange(1, 10).reshape((3,3))
num2 = np.arange(11, 20).reshape((3, 3))
print(num1)
print(num2)
```

```python
sum = num1 + num2
print(sum)

sub = num1 - num2
print(sub)

mul = num1 * num2
print(mul)

div = num1 / num2
print(div)

sqr = num1 ** num2
print(sqr)

root = num2 ** 0.5
print(root)

val = abs(sub)
print(val)
```

```python
rem = num2 % num1
print(rem)
```

```python
root = 2 ** num2
print(root)
```

```python
scaler = num2 + 10
print(scaler)
```

```python
scaler = num2 * 10
print(scaler)
```

```python
array = np.arange(1, 10).reshape(3, 3)
print(array)
```
